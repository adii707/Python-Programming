```python
#Write a program to extract numbers divisible by 2 or 3 and calculate their squares.
l= [2, 3, 5, 6, 8, 9, 10, 12, 15]

result = []

for i in l:
    if i % 2 == 0 or i % 3 == 0:
        square=i*i
        result.append(square)

print("The Result Will Be:", result)
```

```python
#Given a list of strings,extract strings with greater than 3 and convert them into uppercase
l = ["cat", "apple", "dog", "banana", "book"]

result = []

for i in l:
    if len(i) > 3:
        result.append(i.upper())

print("Result:", result)
```

```python
#Given a list of integers, add 10 to every element and then extract elements greater than 30.
l = [10, 15, 20, 25, 30]

new = []

for i in l:
    new.append(i + 10)

result = []

for i in new:
    if i > 30:
        result.append(i)

print("After adding 10:", new)
print("Greater than 30:", result)
```

```python
#Write a program to extract negative numbers and convert them into their absolute values.
l = [-5, 10, -15, 20, -25, 30]

result = []

for i in l:
    if i < 0:
        result.append(abs(i))

print("Result:", result)
```

```python
#Given a list of words, extract words ending with the letter 'n' and calculate their lengths.
l = ["lion", "tiger", "elephant", "monkey", "garden", "sun"]

result = []

for i in l:
    if i.endswith("n"):
        result.append(len(i))

print("Result:", result)
```

```python
#Write a program to extract numbers that are perfect squares from a given list.
l = [1, 4, 9, 10, 16, 20, 25, 30, 36, 50]

result = []

for i in l:
    j = int(i ** 0.5)
    if j * j == i:
        result.append(i)

print("Perfect squares:", result)
```
